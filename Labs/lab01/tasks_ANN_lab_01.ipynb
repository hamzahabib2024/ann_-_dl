##Lab Task 1: Lists, Dictionaries and Tuples

The purpose of this task is to create and manipulate the basic Python data
structures: lists, dictionaries and tuples.

###1. List Operations

```python
# Create the list of colours
colors = ["red", "green", "blue", "yellow", "purple"]
print("Original list      :", colors)

# Add "orange" at the end of the list
colors.append("orange")
print("After append       :", colors)

# Insert "pink" at the second position (index 1)
colors.insert(1, "pink")
print("After insert       :", colors)

# Remove "blue" from the list
colors.remove("blue")
print("After remove       :", colors)

# Reverse the list in place
colors.reverse()
print("After reverse      :", colors)

# Sort the list in alphabetical order
colors.sort()
print("After sort         :", colors)

# Slice the list to get the last three elements
print("Last three elements:", colors[-3:])

# Print the length of the list
print("Length of the list :", len(colors))
```

###2. Dictionary Operations

```python
# Create the dictionary
person = {"name": "John", "age": 25, "city": "New York"}
print("Original dictionary:", person)

# Add a new key-value pair for the email
person["email"] = "john.doe@example.com"
print("After adding email :", person)

# Update the age to 26
person["age"] = 26
print("After updating age :", person)

# Remove the "city" key from the dictionary
del person["city"]          # person.pop("city") does the same job
print("After removing city:", person)

# Check whether the key "name" exists
print("Is 'name' a key?   :", "name" in person)

# Print all the keys and all the values
print("Keys               :", list(person.keys()))
print("Values             :", list(person.values()))

# Print every key together with its value
for key, value in person.items():
    print(f"{key} -> {value}")
```

###3. Tuple Operations

```python
# Create the tuple
numbers = (10, 20, 30, 40, 50)
print("Tuple            :", numbers)

# Access and print the third element (index 2)
print("Third element    :", numbers[2])

# Try to change the second element -> tuples are immutable, so this fails
try:
    numbers[1] = 99
except TypeError as error:
    print("Error while changing an element:", error)

# Create a new tuple by concatenating the existing tuple with (60, 70, 80)
extended = numbers + (60, 70, 80)
print("Concatenated tuple:", extended)

# Unpack the original tuple into individual variables
a, b, c, d, e = numbers
print("Unpacked values  :", a, b, c, d, e)
```

###4. Combining Data Structures

```python
# A dictionary whose values are lists
food = {
    "fruits": ["apple", "banana", "cherry"],
    "vegetables": ["carrot", "broccoli", "spinach"],
}
print("Original dictionary:", food)

# Add a new fruit to the list of fruits
food["fruits"].append("orange")

# Remove "spinach" from the list of vegetables
food["vegetables"].remove("spinach")

# Print the updated dictionary
print("Updated dictionary :", food)

# A list of tuples: each tuple holds a fruit and its colour
fruit_colors = [("apple", "red"), ("banana", "yellow"), ("cherry", "red")]
print("List of tuples     :", fruit_colors)

# Access and print the colour of "banana"
for fruit, color in fruit_colors:
    if fruit == "banana":
        print("Colour of banana   :", color)
```

##Lab Task 2: Functions and Classes

The purpose of this task is to define and use functions, and to design classes
that encapsulate data and behaviour.

###1. Function Definition and Usage

```python
def greet(name):
    """Print a greeting message for the given name."""
    print(f"Hello, {name}! Welcome to the Artificial Neural Networks lab.")


def fibonacci(n):
    """Return the nth Fibonacci number (fibonacci(0) = 0, fibonacci(1) = 1)."""
    if n < 0:
        raise ValueError("n must be a non-negative integer")
    previous, current = 0, 1
    for _ in range(n):
        previous, current = current, previous + current
    return previous


def is_prime(number):
    """Return True if number is prime, False otherwise."""
    if number < 2:
        return False
    # It is enough to test the divisors up to the square root of the number
    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1
    return True


def factorial(n):
    """Return n! computed recursively."""
    if n < 0:
        raise ValueError("n must be a non-negative integer")
    if n in (0, 1):
        return 1
    return n * factorial(n - 1)


def reverse_string(text):
    """Return the reversed copy of the given string."""
    return text[::-1]


# Test every function
greet("Muhammad Hamza Habib")
print("First 10 Fibonacci numbers:", [fibonacci(i) for i in range(10)])
print("fibonacci(10)             :", fibonacci(10))
print("Primes below 20           :", [n for n in range(20) if is_prime(n)])
print("is_prime(29)              :", is_prime(29))
print("factorial(5)              :", factorial(5))
print("reverse_string('Python')  :", reverse_string("Python"))
```

###2. Class Definition and Usage

```python
class Person:
    """A person described by a name, an age and an email address."""

    def __init__(self, name, age, email):
        self.name = name
        self.age = age
        self.email = email

    def introduce(self):
        """Print an introduction message built from the attributes."""
        print(f"Hi, I am {self.name}, I am {self.age} years old. "
              f"You can reach me at {self.email}.")


class Student(Person):
    """A student is a person who also has a student id."""

    def __init__(self, name, age, email, student_id):
        # Reuse the initialisation of the parent class
        super().__init__(name, age, email)
        self.student_id = student_id

    def study(self):
        """Print a message showing that the student is studying."""
        print(f"{self.name} (ID: {self.student_id}) is studying.")


class Course:
    """A course with a name and a list of enrolled Student objects."""

    def __init__(self, course_name):
        self.course_name = course_name
        self.students = []

    def add_student(self, student):
        """Add one Student object to the course."""
        self.students.append(student)
        print(f"{student.name} has been added to '{self.course_name}'.")

    def list_students(self):
        """Print the names of all the students enrolled in the course."""
        print(f"Students enrolled in '{self.course_name}':")
        for number, student in enumerate(self.students, start=1):
            print(f"  {number}. {student.name}")


print("Classes Person, Student and Course have been defined.")
```

###3. Combining Functions and Classes

```python
# Create the course
python_course = Course("Introduction to Python")

# Create three students
student1 = Student("Ali Khan", 20, "ali.khan@example.com", "FA22-BCS-001")
student2 = Student("Sara Ahmed", 21, "sara.ahmed@example.com", "FA22-BCS-002")
student3 = Student("Bilal Iqbal", 22, "bilal.iqbal@example.com", "FA22-BCS-003")

# Add the students to the course
for student in (student1, student2, student3):
    python_course.add_student(student)

print()
# List all the students of the course
python_course.list_students()

print()
# Introduce every student (introduce is inherited from Person)
for student in python_course.students:
    student.introduce()

print()
# Demonstrate the study method for one of the students
student2.study()
```

##Lab Task 3: File I/O with the pickle Module

The purpose of this task is to serialize (pickle) and deserialize (unpickle)
Python objects and to store them in files.

###1. Understanding Pickling

**What is pickling?**
Pickling is the process of converting a Python object (a list, a dictionary, a
class instance, ...) into a stream of bytes so that it can be written to a file
or sent over a network. The opposite process, rebuilding the original object
from those bytes, is called *unpickling*. In Python this is done with the
built-in `pickle` module: `pickle.dump()` writes an object to a file and
`pickle.load()` reads it back.

**Why is it used?**
* It saves the state of a program so that the same data can be used again later.
* It keeps the exact Python types: a list of objects comes back as a list of
  objects, not as plain text that has to be parsed again.
* It can store objects of user defined classes, which formats such as plain
  text or CSV cannot do directly.
* It is useful in machine learning to store trained models, datasets and
  pre-processing objects.

**Note:** pickle files must be opened in binary mode (`"wb"` / `"rb"`), and a
pickle file should only be loaded when it comes from a trusted source, because
unpickling can execute arbitrary code.

###2. Basic Pickling and Unpickling

```python
import pickle

# A list of dictionaries, each dictionary represents one student
students = [
    {"name": "Alice", "age": 20, "grade": "A"},
    {"name": "Bob", "age": 22, "grade": "B"},
    {"name": "Charlie", "age": 19, "grade": "C"},
]

# Serialize (pickle) the list and save it into students.pkl
with open("students.pkl", "wb") as file:
    pickle.dump(students, file)
print("The list of dictionaries has been saved to students.pkl")


def load_students(file_name="students.pkl"):
    """Deserialize students.pkl and return the list of student dictionaries."""
    with open(file_name, "rb") as file:
        return pickle.load(file)


# Load the data back and print it to verify
loaded_students = load_students()
print("Loaded students:")
for student in loaded_students:
    print(" ", student)
```

###3. Working with Custom Classes

```python
class Student:
    """A student described by a name, an age and a grade.

    This class belongs to Task 3 and replaces the Student class of Task 2.
    """

    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def __repr__(self):
        # __repr__ makes the objects readable when a list of them is printed
        return f"Student(name='{self.name}', age={self.age}, grade='{self.grade}')"


# Convert the previous list of dictionaries into a list of Student objects
student_objects = [Student(d["name"], d["age"], d["grade"]) for d in students]

# Serialize the list of objects into students_class.pkl
with open("students_class.pkl", "wb") as file:
    pickle.dump(student_objects, file)
print("The list of Student objects has been saved to students_class.pkl")


def load_student_objects(file_name="students_class.pkl"):
    """Deserialize students_class.pkl and return the list of Student objects."""
    with open(file_name, "rb") as file:
        return pickle.load(file)


# Load the objects back and print them to verify
loaded_objects = load_student_objects()
print("Loaded student objects:")
for student in loaded_objects:
    print(" ", student)
```

###4. Advanced Operations

```python
def add_student(file_name, student):
    """Add a Student object to the list stored in the pickle file."""
    students_list = load_student_objects(file_name)   # read the current list
    students_list.append(student)                     # add the new student
    with open(file_name, "wb") as file:               # write the list back
        pickle.dump(students_list, file)
    print(f"{student.name} has been added to {file_name}")
    return students_list


def remove_student(file_name, student_name):
    """Remove the student with the given name from the pickle file."""
    students_list = load_student_objects(file_name)
    # Keep only the students whose name is different from student_name
    remaining = [s for s in students_list if s.name != student_name]
    if len(remaining) == len(students_list):
        print(f"No student named {student_name} was found in {file_name}")
    else:
        print(f"{student_name} has been removed from {file_name}")
    with open(file_name, "wb") as file:
        pickle.dump(remaining, file)
    return remaining


# Demonstrate add_student: add a new student to students_class.pkl
new_student = Student("Diana", 21, "A")
add_student("students_class.pkl", new_student)
print("After adding    :", load_student_objects("students_class.pkl"))

print()
# Demonstrate remove_student: remove an existing student
remove_student("students_class.pkl", "Bob")
print("After removing  :", load_student_objects("students_class.pkl"))
```
